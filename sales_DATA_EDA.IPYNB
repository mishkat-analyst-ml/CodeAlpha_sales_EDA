```python
import pandas as pd
import matplotlib.pyplot as plt
#Now First i load the data
df= pd.read_csv('sales_data.csv')
```

```python
#Now if we see sales date is in str so i will convert it into datetime format
#first remove the strip white space
df.columns = df.columns.str.strip()
df['Sale_Date'] = pd.to_datetime(df['Sale_Date'])
df.dtypes


```

```python
#checking for null values
df.isnull().sum()
```

```python
#sales analysis
sales_by_region = df.groupby('Region')['Sales_Amount'].sum()
sales_by_region.plot(kind='pie')
plt.title('Total Sales by Region')
plt.xlabel('Region')
plt.ylabel('Total Sales Amount')
plt.show()
```

```python
#Product_category analysis
sales_by_product = df.groupby('Product_Category')['Sales_Amount'].sum()
sales_by_product.plot(kind='bar')
plt.title('Total Sales by Product Category')
plt.xlabel('Product Category')
plt.ylabel('Total Sales Amount')
plt.show()
```

```python
#sales over time
sales_over_time = df.groupby('Sale_Date')['Sales_Amount'].sum()
sales_over_time.plot(kind='line')
plt.title('Sales Over Time')
plt.xlabel('Sale Date')
plt.ylabel('Total Sales Amount')
plt.show()
```

```python
#payment method analysis
sales_by_payment = df.groupby('Payment_Method')['Sales_Amount'].sum()
sales_by_payment.plot(kind='bar')
plt.title('Total Sales by Payment Method')
plt.xlabel('Payment Method')
plt.ylabel('Total Sales Amount')
plt.show()
```

```python
#Discount analysis
sales_by_discount = df.groupby('Discount')['Sales_Amount'].sum()
sales_by_discount.plot(kind='bar')
plt.title('Total Sales by Discount')
plt.xlabel('Discount')
plt.ylabel('Total Sales Amount')
plt.show()
```
